```python
#https://youtu.be/1tAdi6nTSiI

#Nome: Filipe Scal de Araujo
#RM: 569175

#Nome: Gabriel de Medeiros Madureira
#RM: 570297

#Nome: Luis Gustavo Leonart Evangelista
#RM: 572167

Nome: Gabriel de Medeiros Madureira
RM: 570297

Nome: Luis Gustavo Leonart Evangelista
RM: 572167

ciclo = int(input("Quantos ciclos serão feitos: "))

contador_de_ciclos = 0

#Lista da Preção Iterna dod Modúlo
p = []

#Lista de Erros dos Senrores
e = []

#Listas S1
S1_v = []
S1_t = []
S1_d = []

#Listra S2
S2_v = []
S2_t = []
S2_d = []

#Listra S3
S3_v = []
S3_t = []
S3_d = []

#Listra S4
S4_v = []
S4_t = []
S4_d = []

#Listra S5
S5_v = []
S5_t = []
S5_d = []

while contador_de_ciclos < ciclo:
    contador_de_ciclos += 1

    print(f"\n---CICLO {contador_de_ciclos}---")

    #Entradad de Dados Do Sensor 1
    v1 = float(input("Vibração do Sensor 1: "))
    t1 = float(input("Temperatura do Sensor 1: "))
    d1 = float(input("Deformação do Sensor 1: "))

    S1_v.append(v1)
    S1_t.append(t1)
    S1_d.append(d1)

    #Entradad de Dados Do Sensor 2
    v2 = float(input("Vibração do Sensor 2: "))
    t2 = float(input("Temperatura do Sensor 2: "))
    d2 = float(input("Deformação do Sensor 2: "))

    S2_v.append(v2)
    S2_t.append(t2)
    S2_d.append(d2)

    #Entradad de Dados Do Sensor 3
    v3 = float(input("Vibração do Sensor 3: "))
    t3 = float(input("Temperatura do Sensor 3: "))
    d3 = float(input("Deformação do Sensor 3: "))

    S3_v.append(v3)
    S3_t.append(t3)
    S3_d.append(d3)

    #Entradad de Dados Do Sensor 4
    v4 = float(input("Vibração do sensor 4: "))
    t4 = float(input("Temperatura do Sensor 4: "))
    d4 = float(input("Deformação do Sensor 4: "))

    S4_v.append(v4)
    S4_t.append(t4)
    S4_d.append(d4)

    #Entradad de Dados Do Sensor 5
    v5 = float(input("Vibração do sensor 5: "))
    t5 = float(input("Temperatura do Sensor 5: "))
    d5 = float(input("Deformação do Sensor 5: "))

    S5_v.append(v5)
    S5_t.append(t5)
    S5_d.append(d5)

    #Entrada de Dados de Preção
    pp = float(input("Preção Interna do Módulo: "))
    p.append(pp)

    #Detecção de Condições Criticas de Vibração
    if 5.0 < v1 or v1 < -5.0:
      e.append(["Vibaração", "Sensor 1", v1, contador_de_ciclos, "g"])
    if 5.0 < v2 or v2 < -5.0:
      e.append(["Vibaração" ,"Sensor 2", v2, contador_de_ciclos, "g"])
    if 5.0 < v3 or v3 < -5.0:
      e.append(["Vibaração", "Sensor 3", v3, contador_de_ciclos, "g"])
    if 5.0 < v4 or v4 < -5.0:
      e.append(["Vibaração", "Sensor 4", v4, contador_de_ciclos, "g"])
    if 5.0 < v5 or v5 < -5.0:
      e.append(["Vibaração", "Sensor 5", v5, contador_de_ciclos, "g"])

    #Detecção de Condições Criticas de Temperatura
    if -150 > t1 or t1 > 120:
      e.append(["Temperatura", "Sensor 1", t1, contador_de_ciclos, "°C"])
    if -150 > t2 or t2 > 120:
      e.append(["Temperatura", "Sensor 2", t2, contador_de_ciclos, "°C"])
    if -150 > t3 or t3 > 120:
      e.append(["Temperatura", "Sensor 3", t3, contador_de_ciclos, "°C"])
    if -150 > t4 or t4 > 120:
      e.append(["Temperatura", "Sensor 4", t4, contador_de_ciclos, "°C"])
    if -150 > t5 or t5 > 120:
      e.append(["Temperatura", "Sensor 5", t5, contador_de_ciclos, "°C"])

    #Detecção de Condições Criticas de Deformação
    if d1 > 2.5:
      e.append(["Deformação", "Sensor 1", d1, contador_de_ciclos, "mm"])
    if d2 > 2.5:
      e.append(["Deformação", "Sensor 2", d2, contador_de_ciclos, "mm"])
    if d3 > 2.5:
      e.append(["Deformação", "Sensor 3", d3, contador_de_ciclos, "mm"])
    if d4 > 2.5:
      e.append(["Deformação", "Sensor 4", d4, contador_de_ciclos, "mm"])
    if d5 > 2.5:
      e.append(["Deformação", "Sensor 5", d5, contador_de_ciclos, "mm"])

    #Detecção de Condições Criticas de Preção
    if pp < 90 or pp > 110:
      e.append(["Pressão", "Módulo", pp, contador_de_ciclos, "kPa"])

print("\n===== RELATÓRIO FINAL =====")

#Mostra os Erros
if len(e) == 0:
  print("Nhenum erro enontrado")

else:
  i = 0

  while i < len(e):
    print(f"Alerta: {e[i][0]:.2f} critica de {e[i][2]:.2f}{e[i][4]:.2f} no {e[i][1]:.2f} (CICLO {e[i][3]:.2f})!!")
    i += 1

#Tirando a Média dos valores de Vibração
todos_v = S1_v + S2_v + S3_v + S4_v + S5_v
media_v = sum(todos_v)/len(todos_v)

#Tirando a Média dos valores de Temperatura
todos_t = S1_t + S2_t + S3_t + S4_t + S5_t
media_t = sum(todos_t)/len(todos_t)

#Tirando a Média ddos valores de Deformação
todos_d = S1_d + S2_d + S3_d + S4_d + S5_d
media_d = sum(todos_d)/len(todos_d)

#Tirando a Média dos valores de Pressão
media_p = sum(p)/len(p)

#Achando o Valor maximo e minimo de Vibração
maior_v = max(todos_v)
menor_v = min(todos_v)

#Achando o Valor maximo e minimo de Temperatura
maior_t = max(todos_t)
menor_t = min(todos_t)

#Achando o Valor maximo e minimo de Deformação
maior_d = max(todos_d)
menor_d = min(todos_d)

#Achando o Valor maximo e minimo de Preção
maior_p = max(p)
menor_p = min(p)

#Cauculando a Porcentagem de erros encontrados
todos = len(todos_v) + len(todos_d) + len(todos_t) + len(p)
porcentagem = (len(e)/todos)*100

#Achando a Mediana de Vibração
todos_v.sort()
meio_v = int(len(todos_v)/2)
if len(todos_v) % 2 == 0:
  mediana_v = (todos_v[meio_v] + todos_v[meio_v - 1])/2
else:
  mediana_v = todos_v[meio_v]

#Achando a Mediana de Deformação
todos_d.sort()
meio_d = int(len(todos_d)/2)
if len(todos_d) % 2 == 0:
  mediana_d = (todos_d[meio_d] + todos_d[meio_d - 1])/2
else:
  mediana_d = todos_d[meio_d]

#Achando a Mediana de Temperatura
todos_t.sort()
meio_t = int(len(todos_t)/2)
if len(todos_t) % 2 == 0:
  mediana_t = (todos_t[meio_t] + todos_t[meio_t - 1])/2
else:
  mediana_t = todos_t[meio_t]

#Achando a Mediana de Preção
p.sort()
meio_p = int(len(p)/2)
if len(p) % 2 == 0:
  mediana_p = (p[meio_p] + p[meio_p - 1])/2
else:
  mediana_p = p[meio_p]

#Variaveis para o uso do While
valores_v = 0
valores_t = 0
valores_d = 0
valores_p = 0

#Variaveis da soma de para o o cauculo de desvio padrão
soma_v = 0
soma_t = 0
soma_d = 0
soma_p = 0

#Cauculando Desvio Padrão Vibração
while valores_v < len(todos_v):
  soma_v += (todos_v[valores_v] - media_v)**2
  valores_v += 1
desvio_v = (soma_v/len(todos_v))**0.5

#Cauculando Desvio Temperatura
while valores_t < len(todos_t):
  soma_t += (todos_t[valores_t] - media_t)**2
  valores_t += 1
desvio_t = (soma_t/len(todos_t))**0.5

#Cauculando Desvio Padrão Deformação
while valores_d < len(todos_d):
  soma_d += (todos_d[valores_d] - media_d)**2
  valores_d += 1
desvio_d = (soma_d/len(todos_d))**0.5

#Cauculando Desvio Temperatura
while valores_p < len(p):
  soma_p += (p[valores_p] - media_p)**2
  valores_p += 1
desvio_p = (soma_p/len(p))**0.5

#Prints dos Resultaod Finais
print(f"Vibração: Média {media_v:.2}, Valor Maximo {maior_v:.2f}, Valor Minimo{menor_v:.2f}\nTemperatura: Média {media_t:.2f}, Valor Maximo {maior_t:.2f}, Valor Minimo{menor_t:.2f}\nDeformação: Média {media_d:.2f}, Valor Maximo {maior_d:.2f}, Valor Minimo{menor_d:.2f}\nPressão: Média {media_p:.2f}, Valor Maximo {maior_p:.2f}, Valor Minimo{menor_p:.2f}")

if porcentagem > 30:
  print(f"Estado Geral em {porcentagem:.2f}%: Riscos Gerais Estrutrais Elevados - Acionar protocolo de emergencia")
elif 30 >= porcentagem > 10:
  print(f"Estado Geral em {porcentagem:.2f}%: Atenção - Monitoramento intesificado recomendado")
elif 10 > porcentagem:
  print(f"Estado Geral em {porcentagem:.2f}%: Normal - Estrutura dentro dos limites de segurança")
```
